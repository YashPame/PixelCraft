```python
import tkinter as tk
from tkinter import messagebox

from tkinter import ttk

root = tk.Tk()

root.title("Data Entry")
root.geometry("500x500")
root.minsize(300,300)
root.maxsize(600,600)

name_label = tk.Label(root, text = "Enter Your Name", font="Arial 16", fg="green", bg="#F54927")
name_label.pack(padx=20, pady=20)
name_entry = tk.Entry(root)
name_entry.pack()

rollno_label = tk.Label(root, text = "Roll No", font="Arial 16", fg="green", bg="#F54927")
rollno_label.pack(padx=20, pady=20)
rollno_entry = tk.Entry(root)
rollno_entry.pack()

branch_label = tk.Label(root, text = "Branch", font="Arial 16", fg="green", bg="#F54927")
branch_label.pack(padx=20, pady=20)
branch_entry = tk.Entry(root)
branch_entry.pack()

gender_label = tk.Label(root, text = "Branch", font="Arial 16", fg="green", bg="#F54927")
gender_label.pack(padx=20, pady=20)
gender_dropdown = ttk.Combobox(
    root,
    values=["Male", "Female"]
)
gender_dropdown.pack()
agree_value = tk.IntVar()

def submitbtn():
    if not agree_value.get():
        messagebox.showerror("Error", "Aggree tnc")
        return
    name = name_entry.get()
    rollno = rollno_entry.get()
    branch = branch_entry.get()
    gender = gender_dropdown.get()
    print(gender)
    if name == "" or rollno == "" or branch == "" or gender not in ["Male", "Female"]:
        messagebox.showerror("Error", "Please fill all the fields")
        return
    else:
        messagebox.showinfo("Submitted", f"Name: {name}\nRoll No: {rollno}\nBranch: {branch}\nGender: {gender}")

def clearbtn():
    name_entry.delete(0, tk.END)
    rollno_entry.delete(0, tk.END)
    branch_entry.delete(0, tk.END)



tnc= ttk.Checkbutton(
    root,
    text="I agree to Terms and Conditions",
    variable=agree_value
)
tnc.pack()

submit = tk.Button(root, text="Submit", command=submitbtn)
submit.pack(padx=20, pady=20)

clear = tk.Button(root, text="clear", command=clearbtn)
clear.pack(padx=20, pady=20)





root.mainloop()
```
